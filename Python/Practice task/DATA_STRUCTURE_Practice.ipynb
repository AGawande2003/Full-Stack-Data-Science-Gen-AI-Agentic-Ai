1. List is defined using square brackets [].

3. Duplicate values are allowed in a list.

5. A list can hold multiple data types.
6. The id() function returns the unique identity (ID) of the list. Example: id(l)
7. clear() removes all elements from the list.
8. copy() creates a copy of the list. Example: l2 = l.copy()
9. del() delets the lsit
10. count() function number of occurances of value in list
11. index() Adds a value at a specific position.
12. extend() Adds multiple values from another list.
13. remove() Removes the first matching value.
14. pop()  Removes and returns an element.
15. pop(index) Removes an element from a specific position
16. index()  Returns the position of a value.
17. Sliceing is alowed in list
